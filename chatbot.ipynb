```python
import random
from datetime import datetime

print("SmartBot: Hello! I am your assistant.")
print("Type 'bye' anytime to stop the chat.\n")

# predefined responses
responses = {
    "greeting": ["Hello!", "Hi there!", "Hey! Nice to meet you."],
    "how_are_you": ["I'm doing well!", "All good here!", "I'm functioning perfectly!"],
    "ai": ["AI means Artificial Intelligence.",
           "AI allows machines to learn and make decisions.",
           "AI is widely used in data science and automation."]
}

def chatbot_reply(message):

    if message in ["hi", "hello", "hey"]:
        return random.choice(responses["greeting"])

    elif "how are you" in message:
        return random.choice(responses["how_are_you"])

    elif "time" in message:
        current = datetime.now().strftime("%H:%M")
        return f"The current time is {current}"

    elif "ai" in message or "artificial intelligence" in message:
        return random.choice(responses["ai"])

    elif "your name" in message:
        return "I am SmartBot, a rule-based chatbot created using Python."

    elif "help" in message:
        return "You can ask me greetings, AI basics, or time."

    else:
        return "Sorry, I couldn't understand that."

while True:
    user_message = input("You: ").lower()

    if user_message == "bye":
        print("SmartBot: Goodbye! Have a great day.")
        break

    reply = chatbot_reply(user_message)
    print("SmartBot:", reply)
```
