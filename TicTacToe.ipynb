```python
import math


board = [" " for _ in range(9)]

human = "X"
ai = "O"

def display():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()

def empty_cells():
    return [i for i in range(9) if board[i] == " "]

def winner(player):
    combos = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]
    for c in combos:
        if board[c[0]] == board[c[1]] == board[c[2]] == player:
            return True
    return False

def game_over():
    return winner(human) or winner(ai) or len(empty_cells()) == 0


def minimax(is_ai_turn):
    if winner(ai):
        return 1
    if winner(human):
        return -1
    if len(empty_cells()) == 0:
        return 0

    if is_ai_turn:
        best_score = -math.inf
        for cell in empty_cells():
            board[cell] = ai
            score = minimax(False)
            board[cell] = " "
            best_score = max(score, best_score)
        return best_score
    else:
        best_score = math.inf
        for cell in empty_cells():
            board[cell] = human
            score = minimax(True)
            board[cell] = " "
            best_score = min(score, best_score)
        return best_score


def ai_move():
    best_val = -math.inf
    move = None

    for cell in empty_cells():
        board[cell] = ai
        move_val = minimax(False)
        board[cell] = " "

        if move_val > best_val:
            best_val = move_val
            move = cell

    board[move] = ai


def human_move():
    while True:
        try:
            pos = int(input("Enter position (1-9): ")) - 1


            if pos < 0 or pos > 8:
                print("Invalid position! Please enter a number between 1 and 9.")
                continue


            if board[pos] == " ":
                board[pos] = human
                break
            else:
                print("Position already occupied.")

        except ValueError:
            print("Invalid input! Please enter a number.")

print("Tic Tac Toe - Human (X) vs AI (O)")
display()

while not game_over():

    human_move()
    display()

    if game_over():
        break

    print("AI is thinking...")
    ai_move()
    display()


if winner(human):
    print("You win!")
elif winner(ai):
    print("AI wins!")
else:
    print("It's a draw!")
```
